```python
import numpy as np
import matplotlib.pyplot as plt

# -------------------------------
# Data seismik contoh
# -------------------------------
np.random.seed(0)
data = np.random.randn(200, 50)

# -------------------------------
# Pengaturan visualisasi
# -------------------------------
cmap = "seismic"      # pilihan: gray, seismic, viridis, plasma
mode = "manual"      # "auto" atau "manual"

vmin = -2.0
vmax = 2.0

# -------------------------------
# Plot seismik
# -------------------------------
plt.figure(figsize=(6, 8))

if mode == "manual":
    plt.imshow(
        data,
        aspect="auto",
        cmap=cmap,
        vmin=vmin,
        vmax=vmax
    )
else:
    plt.imshow(
        data,
        aspect="auto",
        cmap=cmap
    )

plt.title("Plot Seismik")
plt.xlabel("Trace")
plt.ylabel("Waktu")
plt.colorbar(label="Amplitudo")
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# ==============================
# DATA ANOMALI MAGNETIK (SIMULASI)
# ==============================
np.random.seed(0)
nx, ny = 100, 100

anomali_obs = np.random.randn(nx, ny) * 50
anomali_reg = np.random.randn(nx, ny) * 20
anomali_res = anomali_obs - anomali_reg

# ==============================
# PENGATURAN VISUALISASI
# ==============================
cmap = "seismic"        # pilihan: seismic, viridis, plasma, inferno, coolwarm
scale_mode = "manual"  # "auto" atau "manual"

vmin = -50
vmax = 50

save_plot = True       # True untuk simpan gambar

# ==============================
# FUNGSI PLOT
# ==============================
def plot_anomali(data, title, filename):
    plt.figure(figsize=(5,5))

    if scale_mode == "manual":
        plt.imshow(data, cmap=cmap, vmin=vmin, vmax=vmax)
    else:
        plt.imshow(data, cmap=cmap)

    plt.title(title)
    plt.xlabel("X")
    plt.ylabel("Y")
    plt.colorbar(label="Anomali Magnetik (nT)")
    plt.show()

    if save_plot:
        plt.figure(figsize=(5,5))
        plt.imshow(data, cmap=cmap)
        plt.colorbar(label="Anomali Magnetik (nT)")
        plt.savefig(filename, dpi=300)
        plt.close()

# ==============================
# TAMPILKAN SEMUA PETA
# ==============================
plot_anomali(anomali_obs, "Anomali Magnetik Observasi", "anomali_observasi.png")
plot_anomali(anomali_reg, "Anomali Magnetik Regional", "anomali_regional.png")
plot_anomali(anomali_res, "Anomali Magnetik Residual", "anomali_residual.png")

print("Plot selesai dibuat dan disimpan.")
```
